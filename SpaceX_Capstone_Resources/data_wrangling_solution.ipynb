# SpaceX Falcon 9 — Data WranglingPrepares the course launch table, resets the flight index, handles missing payload mass, and creates the binary first-stage landing label.

## Load and filter the data

```python
import pandas as pdimport numpy as npurl = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"df = pd.read_csv(url)df = df[df["BoosterVersion"] == "Falcon 9"].copy()df = df.sort_values("Date").reset_index(drop=True)df["FlightNumber"] = np.arange(1, len(df) + 1)print(df.shape)df.head()
```

## Impute payload mass and define Class

```python
payload_mean = df["PayloadMass"].mean()df["PayloadMass"] = df["PayloadMass"].fillna(payload_mean)# A successful first-stage landing is labelled 1; unsuccessful outcomes are 0.df["Class"] = df["Outcome"].isin(["True ASDS", "True RTLS", "True Ocean"]).astype(int)print("Payload mean used:", round(payload_mean, 2), "kg")print(df["Class"].value_counts().sort_index())
```

```python
df.to_csv("dataset_part_2_recreated.csv", index=False)
```
