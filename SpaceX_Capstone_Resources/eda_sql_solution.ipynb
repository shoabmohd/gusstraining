## Load SQL source data

```python
import sqlite3import pandas as pdurl = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv"df_sql = pd.read_csv(url)con = sqlite3.connect("spacex_capstone.db")df_sql.to_sql("SPACEXTABLE", con, if_exists="replace", index=False)print(df_sql.shape)
```

## Core SQL tasks

```python
queries = { "unique_sites": 'SELECT DISTINCT Launch_Site FROM SPACEXTABLE ORDER BY Launch_Site', "cca_sites": "SELECT * FROM SPACEXTABLE WHERE Launch_Site LIKE 'CCA%' LIMIT 5", "nasa_crs_payload": "SELECT SUM(PAYLOAD_MASS__KG_) AS total_payload_kg FROM SPACEXTABLE WHERE Customer='NASA (CRS)'", "f9_v11_average": "SELECT AVG(PAYLOAD_MASS__KG_) AS avg_payload_kg FROM SPACEXTABLE WHERE Booster_Version='F9 v1.1'", "first_ground_landing": "SELECT MIN(Date) AS first_ground_landing FROM SPACEXTABLE WHERE Landing_Outcome='Success (ground pad)'"}for name, sql in queries.items():    print("\n", name)    display(pd.read_sql_query(sql, con))
```

## Landing outcome queries

```python
sql = """SELECT Booster_Version FROM SPACEXTABLE WHERE Landing_Outcome='Success (drone ship)' AND PAYLOAD_MASS__KG_ > 4000 AND PAYLOAD_MASS__KG_ < 6000"""display(pd.read_sql_query(sql, con))sql = "SELECT Mission_Outcome, COUNT(*) AS n FROM SPACEXTABLE GROUP BY Mission_Outcome ORDER BY n DESC"display(pd.read_sql_query(sql, con))con.close()
```
