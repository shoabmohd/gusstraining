# SpaceX Falcon 9 Launch Records — Web ScrapingExtracts historical launch records from the June 2021 Wikipedia snapshot used in the IBM lab.

## Download and parse the static page

```python
import requestsfrom bs4 import BeautifulSoupimport pandas as pdurl = "https://en.wikipedia.org/w/index.php?title=List_of_Falcon_9_and_Falcon_Heavy_launches&oldid=1027686922"headers = {"User-Agent": "Mozilla/5.0 (compatible; CapstoneStudent/1.0)"}response = requests.get(url, headers=headers, timeout=30)response.raise_for_status()soup = BeautifulSoup(response.text, "html.parser")print(soup.title)
```

## Locate the launch table

```python
tables = soup.find_all("table", class_="wikitable plainrowheaders collapsible")if len(tables) < 3:    raise RuntimeError("Expected launch table was not found.")launch_table = tables[2]headers = [th.get_text(" ", strip=True) for th in launch_table.find_all("th")]print(headers)
```

## Extract rows with numeric flight numbers

```python
records = []for tr in launch_table.find_all("tr"):    th = tr.find("th")    tds = tr.find_all("td")    if th is None or not tds:        continue    flight_no = th.get_text(" ", strip=True)    if flight_no.isdigit():        records.append([flight_no] + [td.get_text(" ", strip=True) for td in tds])web_launches = pd.DataFrame(records)print("Parsed launch rows:", len(web_launches))web_launches.head()
```

```python
web_launches.to_csv("spacex_web_scraped.csv", index=False)
```
