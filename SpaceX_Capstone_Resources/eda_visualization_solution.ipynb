## Load dataset and libraries

```python
import pandas as pdimport seaborn as snsimport matplotlib.pyplot as plturl = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv"df = pd.read_csv(url)print(df.shape)df.head()
```

## Flight number and payload mass by launch site

```python
sns.catplot(data=df, x="FlightNumber", y="LaunchSite", hue="Class", kind="strip", height=5, aspect=1.5)plt.show()sns.catplot(data=df, x="PayloadMass", y="LaunchSite", hue="Class", kind="strip", height=5, aspect=1.5)plt.show()
```

## Success rate by orbit

```python
orbit_success = df.groupby("Orbit", as_index=False)["Class"].mean()sns.barplot(data=orbit_success, x="Orbit", y="Class")plt.ylabel("Average landing success rate")plt.xticks(rotation=45)plt.show()
```
