## Request and parse the launch snapshot

```python
import requestsimport pandas as pdfrom pandas import json_normalizeapi_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/API_call_spacex_api.json"response = requests.get(api_url, timeout=30)response.raise_for_status()launch_json = response.json()launches = json_normalize(launch_json)print("HTTP status:", response.status_code)print("Launch records:", len(launches))launches.head()
```

## Select and export relevant launch fields

```python
selected = [c for c in launches.columns if c in {"name", "date_utc", "success", "rocket", "launchpad", "payloads", "cores"}]launches[selected].head()
```
